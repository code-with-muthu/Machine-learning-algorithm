## Importing the libraries

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

## Importing the dataset 

```python
dataset = pd.read_csv('Ads_CTR_Optimisation.csv')

# in the datset  all the adds has different conversion rate (ie. ad1 -different coversion rate , ad2 -- different conersion rate  .... here each ad has different click through rate )
```

## implementing thomson sampling

```python
import random
N = 10000 # number of user in a dataset 
d = 10 # number of adds in a dataset 
ads_selected =[]

numbers_of_rewards_1= [0] * d 
numbers_of_rewards_0= [0]*d 
total_reward = 0

for n in range(0,N):
    ad = 0 
    max_random = 0 
    for i in range(0,d):
        random_beta = random.betavariate(numbers_of_rewards_1[i]+1, numbers_of_rewards_0[i]+1) # random draw of add (i)
        if (random_beta > max_random):
            max_random = random_beta # updating if the values are higher 
            ad = i



            # updating the vaiables 
    ads_selected.append(ad)
    reward = dataset.values[n,ad]
    if reward ==1:
        numbers_of_rewards_1[ad] = numbers_of_rewards_1[ad]+1
    else:
        numbers_of_rewards_0[ad] = numbers_of_rewards_0[ad]+1
    total_reward =total_reward + reward


```

## visulaizing the results 

```python
plt.hist(ads_selected)
plt.title('histogrms ofthe adds selected')
plt.xlabel('Ads')
plt.ylabel('number of time each ad was selected ')
plt.show()
```
