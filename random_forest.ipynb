```python
# # What is Random Forest?

# A Random Forest is a supervised machine learning algorithm used for:

# Classification → predicting categories
# Example: Spam or Not Spam
# Regression → predicting numerical values
# Example: House price prediction

# It is called Random Forest because it creates many Decision Trees and combines their results to make a better prediction.

# Basic Idea

# Instead of using one Decision Tree, Random Forest uses:

# 🌳 Tree 1
# 🌳 Tree 2
# 🌳 Tree 3
# 🌳 Tree 4
# ... many trees together

# Then it takes the final decision based on all trees.

# How Random Forest Works
# Step 1: Takes Random Data

# The algorithm creates multiple subsets from the original dataset using random sampling.

# Example:

# Student	Hours Studied	Result
# A	2	Fail
# B	5	Pass
# C	8	Pass

# Different trees get different random portions of data.

# Step 2: Creates Multiple Decision Trees

# Each tree learns differently because:

# data is random
# features are random

# Example:

# Tree 1:

# Hours > 4 ? → Pass

# Tree 2:

# Attendance > 70% ? → Pass

# Tree 3:

# Assignments completed ? → Pass
# Step 3: Final Prediction
# For Classification

# Uses Majority Voting

# Example:

# Tree	Prediction
# Tree 1	Pass
# Tree 2	Pass
# Tree 3	Fail

# Final Output:
# ✅ Pass

# For Regression

# Uses Average

# Example:

# Tree	Salary Prediction
# Tree 1	50,000
# Tree 2	55,000
# Tree 3	53,000

# Final Output:

# Average = 52,666
# Why Random Forest is Better than Decision Tree
# Decision Tree	Random Forest
# Uses one tree	Uses many trees
# Can overfit easily	Reduces overfitting
# Less accurate	More accurate
# Sensitive to data	More stable
# Advantages of Random Forest

# ✅ High accuracy
# ✅ Handles large datasets
# ✅ Reduces overfitting
# ✅ Works for classification and regression
# ✅ Handles missing values well
# ✅ Good feature importance detection

# Disadvantages

# ❌ Slower than Decision Tree
# ❌ Uses more memory
# ❌ Harder to visualize
# ❌ Training takes longer

# Important Concepts
# 1. Ensemble Learning

# Combining multiple models to improve performance.

# Random Forest = collection of Decision Trees.

# 2. Bagging (Bootstrap Aggregation)

# Random Forest uses:

# random data sampling
# multiple trees
# averaging/voting

# This technique is called Bagging.

# Example Real-World Applications
# Classification
# Spam email detection
# Disease prediction
# Fraud detection
# Regression
# Stock price prediction
# House price prediction
# Sales forecasting
# Simple Flow of Random Forest
# Dataset
#    ↓
# Random Sampling
#    ↓
# Multiple Decision Trees
#    ↓
# Each Tree Predicts
#    ↓
# Voting / Averaging
#    ↓
# Final Prediction




# Scikit-Learn Example (Python)
# Classification
# from sklearn.ensemble import RandomForestClassifier

# model = RandomForestClassifier(n_estimators=100)

# model.fit(X_train, y_train)

# prediction = model.predict(X_test)
# Regression
# from sklearn.ensemble import RandomForestRegressor

# model = RandomForestRegressor(n_estimators=100)

# model.fit(X_train, y_train)

# prediction = model.predict(X_test)
# Important Parameter
# n_estimators

# Number of trees in the forest.

# Example:

# RandomForestClassifier(n_estimators=100)

# More trees:

# better accuracy
# but slower training
```

# R- squared method

```python
# What is R-Squared?

# R-Squared (also called the Coefficient of Determination) is an evaluation metric used mainly in Regression models.

# It measures:

# How well the model explains the variation in the target variable.

# In simple words:

# 👉 It tells us how good the regression model fits the data.

# Simple Meaning

# Suppose you are predicting:

# House prices
# Salary
# Stock prices

# R² tells:

# how much of the actual data is explained by the model.
# R-Squared Formula

# R
# 2
# =1−
# SS
# tot
# 	​

# SS
# res
# 	​

# 	​


# Where:

# SS
# res
# 	​

#  = Residual Sum of Squares
# (prediction errors)
# SS
# tot
# 	​

#  = Total Sum of Squares
# (total variation in data)
# Understanding the Values
# 1. R
# 2
# =1

# ✅ Perfect prediction
# Model explains all variation.

# Example:

# R² = 1.0 → 100% accurate fit
# 2. R
# 2
# =0

# ❌ Model explains nothing.

# The model is no better than taking the mean.

# 3. 0<R
# 2
# <1

# Partial explanation.

# Example:

# R² = 0.85

# Means:

# model explains 85% of data variation
# 15% remains unexplained
# 4. Negative R²

# ❌ Very poor model

# Happens when predictions are worse than simply predicting the mean.

# Example

# Suppose:

# Hours Studied	Actual Marks	Predicted Marks
# 2	40	42
# 4	50	48
# 6	65	67

# If predictions are close to actual values:

# ✅ R² will be high.

# Graphical Understanding
# High R²

# Points are close to regression line.

# • • • • •
#   Regression Line
# Low R²

# Points are far from regression line.

# •     •
#     •
#        •
# Why R-Squared is Important

# It helps to:

# evaluate regression performance
# compare models
# understand model fit
# Advantages

# ✅ Easy to understand
# ✅ Good for regression evaluation
# ✅ Helps compare models

# Limitations

# ❌ Cannot detect overfitting alone
# ❌ Higher R² does not always mean better model
# ❌ Not suitable for classification problems

# Adjusted R-Squared

# When more features are added, normal R² may increase even if features are useless.

# So we use:

# Adjusted R²

# It penalizes unnecessary features.

# Difference Between R² and Adjusted R²
# R²	Adjusted R²
# Always increases with features	Penalizes useless features
# Simpler metric	More reliable
# Used for basic evaluation	Used in multiple regression
# Python Example
# from sklearn.metrics import r2_score

# r2 = r2_score(y_test, y_pred)

# print(r2)
```
