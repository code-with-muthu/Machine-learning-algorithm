Why is it called “Naive”?

It is called Naive because it assumes that:

All input features are independent of each other.

Example:
While detecting spam emails, words like:

“free”
“offer”
“discount”

This assumption is simple (“naive”), but the algorithm works surprisingly well.

Likelihood in Machine Learning

Likelihood means:

The probability of observing the given data assuming a particular model or class is true.

In simple words:

“If this class/model is correct, how likely are we to get this data?”

In Bayes’ Theorem
P(A∣B)=
P(B)
P(B∣A)P(A)
	​

P(A)
P(B∣A)
P(B∣¬A)
P(A∣B)=
P(B)
P(B∣A)P(A)
	​

≈0.68,P(B)≈0.25
P(B)=0.25
P(B|A)P(A)=0.17
P(A|B)~0.68
Posterior = useful evidence / total evidence

Here:

P(B∣A)

is called the Likelihood.

It means:

Probability of observing evidence B when event/class A is true.

Simple Example

Suppose:

We want to know whether an email is Spam.
The word “Offer” appears in the email.

Now:

P(Offer∣Spam)

means:

Probability that the word “Offer” appears given that the email is Spam.

This is the likelihood.

Real-Life Analogy

Imagine:

A doctor tests for a disease.
A patient has symptoms like fever.

Likelihood asks:

“If the patient truly has the disease, what is the probability of seeing these symptoms?”

In Machine Learning

Likelihood helps the model determine:

Which class/model makes the observed data most probable.

Algorithms using likelihood:

Naive Bayes
Logistic Regression
Maximum Likelihood Estimation (MLE)
Difference Between Probability and Likelihood
Probability	Likelihood
Predicts future outcomes	Evaluates model based on observed data
Data may vary	Model/class may vary
Example: Probability of rain tomorrow	Example: How likely this weather pattern indicates rain
Key Idea
Probability → Chance of an event happening
Likelihood → How well a model/class explains observed data
One-Line Definition

Likelihood is the probability of observed data given that a particular hypothesis, class, or model is true. -->


### Importing the libraries

```python
import numpy as np
import matplotlib.pyplot as plt 
import pandas as pd 
```

### importing the datasets

```python
dataset = pd.read_csv('Social_Network_Ads.csv')
x=dataset.iloc[:, :-1].values
y=dataset.iloc[:, -1].values
```

### splitting the dataset into train and test 

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.25, random_state = 0)
```

```python
print(x_train)
```

```python
print(y_train)
```

```python
print(y_test)
```

```python
print(x_test)
```

### feature scaling 

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train =sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

```python
print(x_train)
```

```python
print(y_train)
```

```python
print(x_test)
```

```python
print(y_test)
```

### training the naive bayes model on training set

```python
from sklearn.tree import DecisionTreeClassifier
dtc_obj= DecisionTreeClassifier(criterion='entropy',random_state=0)
#train
dtc_obj.fit(x_train,y_train)
```

### predicting new result 

```python
print(dtc_obj.predict(sc.transform([[30,87000]])))
```

### predicting the test set result 

```python
y_pred = dtc_obj.predict(x_test)
# since we are dooing with intgers . not required np.set_printoptions(precision = 2)
print(np.concatenate((y_pred.reshape(len(y_pred),1), y_test.reshape(len(y_test),1)),1))
```

### making confusion matrix

```python
from sklearn.metrics import confusion_matrix, accuracy_score
cm_obj = confusion_matrix(y_test, y_pred)
print(cm_obj)


# to check accuracy 
as_obj = accuracy_score(y_test,y_pred) # checks accuracy 
print(as_obj)
```

### visualizing the training set results 

```python
from matplotlib.colors import ListedColormap


# Inverse transform
x_set, y_set = sc.inverse_transform(x_train), y_train

# Create meshgrid with larger step size
x1, x2 = np.meshgrid(
    np.arange(start=x_set[:, 0].min() - 1,
              stop=x_set[:, 0].max() + 1,
              step=0.25),

    np.arange(start=x_set[:, 1].min() - 1000,
              stop=x_set[:, 1].max() + 1000,
              step=100)
)

# Predict values
predictions = dtc_obj.predict(
    sc.transform(np.array([x1.ravel(), x2.ravel()]).T)
).reshape(x1.shape)

# Plot contour
plt.contourf(
    x1,
    x2,
    predictions,
    alpha=0.75,
    cmap=ListedColormap(('red', 'green'))
)

# Axis limits
plt.xlim(x1.min(), x1.max())
plt.ylim(x2.min(), x2.max())

# Scatter points
for i, j in enumerate(np.unique(y_set)):
    plt.scatter(
        x_set[y_set == j, 0],
        x_set[y_set == j, 1],
        c=ListedColormap(('red', 'green'))(i),
        label=j
    )

# Labels
plt.title('Decision tree - classification(Training set)')
plt.xlabel('Age')
plt.ylabel('Estimated Salary')
plt.legend()

plt.show()
```

### visualizing the test set results 

```python
from matplotlib.colors import ListedColormap
import numpy as np
import matplotlib.pyplot as plt

# Inverse transform
x_set, y_set = sc.inverse_transform(x_test), y_test

# Create meshgrid with SAFE step sizes
x1, x2 = np.meshgrid(
    np.arange(
        start=x_set[:, 0].min() - 1,
        stop=x_set[:, 0].max() + 1,
        step=0.5
    ),

    np.arange(
        start=x_set[:, 1].min() - 1000,
        stop=x_set[:, 1].max() + 1000,
        step=500
    )
)

# Predict values
grid_points = np.array([x1.ravel(), x2.ravel()]).T

predictions = dtc_obj.predict(
    sc.transform(grid_points)
).reshape(x1.shape)

# Plot decision boundary
plt.contourf(
    x1,
    x2,
    predictions,
    alpha=0.75,
    cmap=ListedColormap(('red', 'green'))
)

# Axis limits
plt.xlim(x1.min(), x1.max())
plt.ylim(x2.min(), x2.max())

# Plot test points
for i, j in enumerate(np.unique(y_set)):
    plt.scatter(
        x_set[y_set == j, 0],
        x_set[y_set == j, 1],
        c=ListedColormap(('red', 'green'))(i),
        label=j
    )

# Labels
plt.title('Decision tree - classification(Test set)')
plt.xlabel('Age')
plt.ylabel('Estimated Salary')
plt.legend()

plt.show()
```
