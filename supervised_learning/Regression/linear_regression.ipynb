## Simple linear regression 

### importing libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

### importing dataset 

```python
dataset=pd.read_csv('Salary_Data.csv')

x = dataset.iloc[:, :-1].values # taken alll the rows

y=dataset.iloc[:,-1].values

print(x,y)
```

### splitting the dataset into training and test dataset 

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2, random_state=0)
```

## training the simple linear regression model on the training set

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()

# to train a regression model -- use fit()
lr.fit(x_train,y_train)  
```

### Predicting the test set results 

```python
# to predict we use predict()
y_pred=lr.predict(x_test) 
# y pred contains the predicted resutls and y test contains the actaual results 
# x test has the test ting geatures and the output is compard eith y tst 
```

### visualizing the training set results

```python
plt.scatter(x_train,y_train,color= 'red')
plt.plot(x_train,lr.predict(x_train),color = 'blue')  # regression line
plt.title('salary vs experience(training set)')
plt.xlabel('years of experience')
plt.ylabel('salary')
plt.show()
```

### visualising the test set results 

```python
plt.scatter(x_test,y_test,color= 'red')

# here no need to change beacause its taken from trained one 
plt.plot(x_train,lr.predict(x_train),color = 'blue')  # regression line
plt.title('salary vs experience(training set)')
plt.xlabel('years of experience')
plt.ylabel('salary')
plt.show()
```
