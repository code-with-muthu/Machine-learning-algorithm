### data preprocessing

##### import libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd 
```

##### importing the dataset 

```python
dataset=pd.read_csv('Data.csv')
# create ntwo new entities - materic or features  and second ----dependent variable vector
# features are the columns using which we are doing to predict the dependent variable or outcome 
# features also called as independent variable 

x = dataset.iloc[:, :-1].values # taken alll the rows
# x- feature column
#iloc= locate index- takes index of row /column which we need to extrct 
# :   --> will take all the rows
#  :-1 ----> python exclude the last values (n), so here exctly sice x is input / featue , it will take all feature columns except the outputcolumn  
# : refers range

y=dataset.iloc[:,-1].values
# we need single column so we removed : (range) from column
print(x,y)
```

##### taking care of missing values

```python
# filling missing values by average 
# using scikit learn to handle missing values 
from sklearn.impute import SimpleImputer
imputer_obj = SimpleImputer(missing_values =np.nan,strategy='mean')
imputer_obj.fit(x[:,1:3])# x numerical data
# 1 -- age , 2 --- sal 

x[:,1:3]=imputer_obj.transform(x[:,1:3]) 
```

```python
print(x)
```

##### encoding categorical data

```python
# coverting category into numbers  using one hot encoding 
#turning country column into three columns with binary values
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

ct = ColumnTransformer(transformers=[('enocoder',OneHotEncoder(),[0])] , remainder='passthrough')
# remainder - passthrough --> this is implemetd because we have age and category , we need in the dataset 
# if its not implemeted it will keep only country  with binary value
#remainder='passthrough' keeps the remaining columns (Age, Salary, Purchased) unchanged.
x= np.array(ct.fit_transform(x))
print(x)
```

## feature scaling

#### wedo feature scaling  so that one feature do not dominate other feature 

```python
# Feature scaling is applied after splitting the dataset to prevent data leakage.

# What is Data Leakage?

# Data leakage happens when information from the test set indirectly influences the training process.

# If scaling is done before splitting, the scaling method calculates values like:

# mean
# standard deviation
# minimum
# maximum

# using the entire dataset, including the test data.

# This gives the model indirect knowledge about the test set, which is incorrect.

# Correct Order
# Import dataset
# Handle missing values
# Encode categorical data
# Split into training and test set
# Apply feature scaling
```

```python
# keeping all the features in same scale 
```

#### how feature scaling is applied or implemented??

##### using 2 methids --> standardisation and normalization
##### standardisation ----> x-mean(x)/sd(x)   range --> can be ussed on all dataset
##### normalization ----> x= x-min(x)/max(x)-min(x)    range 

```python
# standard scalerfrom 
from sklearn.preprocessing import StandardScaler
sdc= StandardScaler() #no arguments neded to pass 

# x -- only numerial values and 3: age and all
x_train[:,3:]=sdc.fit_transform(x_train[:,3:])#ntaking oly numerical columns 

# similar;y doing or x_test , but for x_test we will not  qgqin apply scaler method 
# because it is test set , it should get the scaled values on training only 
# if scaler applied on test, it will create a new scaler value 
# apply same scler on test set which was applied on train set 
x_test[:,3:]=sdc.fit_transform(x_test[:,3:])


# fit = will get mean /average and sd from all features (formula) , and tranfrom will apply the formula so thta they can be at same scale 


print(x_train)
```
