# CLUSTERING


## CONCEPT 

###  DEF :
 Clustering in Machine Learning is an unsupervised learning technique used to group similar data points into clusters based on their characteristics or patterns.

  Unlike classification, clustering does not require labeled data.

### COMMON CLUSTERING ALGORITHMS ARE :
  1. K-Means Clustering
  2. Hierarchical Clustering
  3. DBSCAN
  4. Mean Shift
