
### def :
Association Rule Learning is a machine learning technique used to discover interesting relationships or patterns between items in large datasets.

It is mainly used in:

Market basket analysis
Recommendation systems
Customer behavior analysis
Simple Idea

Association Rule Learning answers questions like:

“If a customer buys item A, what other item is likely to be bought?”

Example:

Customers buying bread often buy butter.

So the rule becomes:

Bread → Butter
Real-Life Example

In supermarkets:

Milk → Bread

means:

Customers who buy milk are also likely to buy bread.

This helps stores:

arrange products
create offers
improve sales
Working of Association Rule Learning
Step 1: Collect Transaction Data

Example dataset:

Transaction ID	Items Purchased
T1	Bread, Butter, Milk
T2	Bread, Jam
T3	Milk, Butter
T4	Bread, Butter

Step 2: Find Frequent Itemsets
The algorithm identifies item combinations that occur frequently.
Example:

Bread & Butter
Milk & Butter

Step 3: Generate Association Rules

Rules are formed like:
Bread → Butter

Meaning:

If Bread is purchased,
Butter is likely purchased too.

Important Terms
1. Support
Support measures how frequently an itemset appears.

Formula:
Support(A→B)=Transactions containing A and B / Total Transactions
Higher support means:rule appears more often.


2. Confidence

Confidence measures how often B is purchased when A is purchased.

Formula:

Confidence(A→B)=
Transactions containing A
Transactions containing A and B
	​


Higher confidence means:

stronger relationship.
3. Lift

Lift measures how strongly two items are related.

Formula:

Lift(A→B)=
Support(B)
Confidence(A→B)
	​


Interpretation:

Lift > 1 → positive association
Lift = 1 → no association
Lift < 1 → negative association
Types of Association Rule Algorithms
1. Apriori Algorithm

Most popular algorithm.

Working:

Finds frequent itemsets
Generates rules step by step
2. Eclat Algorithm

Uses depth-first search for faster processing.

3. FP-Growth Algorithm

Efficient for large datasets.

Example of Apriori Rule

Suppose:

80% of customers buying bread also buy butter

Rule:

Bread → Butter

Confidence = 80%

## Applications

Association Rule Learning is used in:

Market basket analysis
Product recommendation
Cross-selling
Website recommendation
Medical diagnosis
Fraud detection
## Advantages
Finds hidden relationships
Improves business decisions
Useful in recommendation systems
Easy to interpret
## Disadvantages
Computationally expensive for large datasets
Generates many unnecessary rules
Requires careful threshold selection

### importing libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

### Data preprocessing

```python
dataset = pd.read_csv('Market_Basket_Optimisation.csv',header=None) # our dataset does not has any header 


# apriori model expects the dataset in some other format not in pandas , it acepts in list of transaction , so we use for loop to iterate all tge transaction of the dtset 
transactions =[] # creating a empty list 
for i in range (0, 7501):#row
    transactions.append([str(dataset.values[i,j])for j in range (0,20)])# col
    # dataset.values[i,j]) index of row (i) and index of column (j) , but here we cant access the s=datset directly  so we are implementing '.values'
    # and it is a rule that all the elemnts in list should be string , so we have passed  'str'
    
```

### Training the Apriori model on the dataset 

```python
from apyori import apriori
rules_ap_obj = apriori(transactions=transactions,min_support=0.003, min_confidence= 0.2,min_lift = 3, min_length = 2, max_length = 2)
```

### Visualizing the results

### Displaying the first results coming directly from the output of the apriori function

```python
results = list(rules_ap_obj) # high cahnce of buying is predicted/ said using confidence 
results
```

### putting the reults well organised into a Pandas DataFrame 

```python
def inspect(results):
    lhs = [tuple(result[2][0][0])[0] for result in results]
    rhs = [tuple(result[2][0][1])[0] for result in results]
    supports = [result[1] for result in results]
    confidence = [result[2][0][2] for result in results]
    lifts = [result[2][0][3] for result in results]
    return list(zip(lhs, rhs, supports , confidence ,lifts))
resultsinDataFrame = pd.DataFrame(inspect(results),columns = ['left hand side', 'right hand side','support','confidence','lift'])
```

```python
resultsinDataFrame.nlargest(n = 10, columns='lift')
```
