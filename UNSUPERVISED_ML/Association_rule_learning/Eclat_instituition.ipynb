## ECLAT Algorithm in Machine Learning

The ECLAT (Equivalence Class Clustering and bottom-up Lattice Traversal) algorithm is a popular association rule mining algorithm used to find frequent itemsets in a dataset.

It is mainly used in Market Basket Analysis, where we discover items that are frequently purchased together.

Example:

Customers who buy bread may also buy butter.
Customers who buy laptop may also buy mouse.
Main Idea of ECLAT

Unlike the Apriori algorithm, which uses a horizontal database format, ECLAT uses a vertical database format.

Horizontal Format (Apriori)
Transaction ID	Items
T1	Bread, Milk
T2	Bread, Butter
T3	Milk, Butter
Vertical Format (ECLAT)
Item	Transaction IDs (TID Set)
Bread	T1, T2
Milk	T1, T3
Butter	T2, T3

ECLAT finds frequent itemsets by intersecting transaction ID sets.

Working of ECLAT Algorithm
Step 1: Convert Dataset into Vertical Format

Example transactions:

TID	Items
T1	A, B, C
T2	A, C
T3	B, C
T4	A, B

Vertical representation:

Item	TID Set
A	{T1, T2, T4}
B	{T1, T3, T4}
C	{T1, T2, T3}
Step 2: Find Frequent 1-Itemsets

Suppose minimum support = 2.

Support count:

A → 3
B → 3
C → 3

All are frequent.

Step 3: Generate 2-Itemsets Using Intersection
A ∩ B

{T1, T2, T4} ∩ {T1, T3, T4}

= {T1, T4}

Support = 2

So, {A, B} is frequent.

A ∩ C

{T1, T2, T4} ∩ {T1, T2, T3}

= {T1, T2}

Support = 2

So, {A, C} is frequent.

B ∩ C

{T1, T3, T4} ∩ {T1, T2, T3}

= {T1, T3}

Support = 2

So, {B, C} is frequent.

Step 4: Generate 3-Itemsets
A ∩ B ∩ C

{T1, T4} ∩ {T1, T2}

= {T1}

Support = 1

Not frequent if minimum support = 2.

Final Frequent Itemsets
{A}
{B}
{C}
{A, B}
{A, C}
{B, C}
Important Concepts
Support

Support measures how frequently an itemset appears.

Formula:

Support(A)=
Total Transactions
Transactions containing A
	​


Example:
If A appears in 3 out of 4 transactions:

Support(A) = 3/4 = 0.75

Advantages of ECLAT
Faster than Apriori for large datasets.
Uses set intersection operations which are efficient.
Reduces repeated database scanning.
Works well for dense datasets.
Disadvantages of ECLAT
Consumes more memory because TID sets are stored.
Less efficient for sparse datasets.
Recursive processing can become complex for huge datasets.
Difference Between Apriori and ECLAT
Feature	Apriori	ECLAT
Data Format	Horizontal	Vertical
Method	Candidate generation	TID intersection
Speed	Slower	Faster
Database Scans	Multiple	Fewer
Memory Usage	Lower	Higher
Applications of ECLAT
Market basket analysis
Product recommendation systems
Web usage mining
Medical diagnosis analysis
Fraud detection
Customer behavior analysis

### importing the libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

### data preprocessing 

```python
dataset = pd.read_csv('Market_Basket_Optimisation.csv',header=None) # our dataset does not has any header 



transactions =[] 
for i in range (0, 7501):#row
    transactions.append([str(dataset.values[i,j])for j in range (0,20)])# col
    # dataset.values[i,j]) index of row (i) and index of column (j) , but here we cant access the s=datset directly  so we are implementing '.values'
    # and it is a rule that all the elemnts in list should be string , so we have passed  'str'
    
```

### training the eclat model in the dataset 

```python
from apyori import apriori
rules_ap_obj = apriori(transactions=transactions,min_support=0.003, min_confidence= 0.2,min_lift = 3, min_length = 2, max_length = 2)
```

## Visualising the result 

### displying the first results coming directly from the output of the apriori function

```python
results = list(rules_ap_obj) # high cahnce of buying is predicted/ said using confidence 
results
```

### putting the results well organised into a pandas dataFrame

```python
def inspect(results):
    lhs = [tuple(result[2][0][0])[0] for result in results]
    rhs = [tuple(result[2][0][1])[0] for result in results]
    supports = [result[1] for result in results]
    return list(zip(lhs, rhs, supports ))
resultsinDataFrame = pd.DataFrame(inspect(results),columns = ['product 1','product 2','support'])
```
