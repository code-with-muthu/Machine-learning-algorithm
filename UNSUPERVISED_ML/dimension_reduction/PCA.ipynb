PART 1 — WHAT IS PCA?
SIMPLE DEFINITION

PCA (Principal Component Analysis) is an unsupervised machine learning algorithm used to reduce the number of features while preserving the most important information from the dataset.

MAIN IDEA OF PCA

Suppose dataset has:
✅ Many features/columns

Some features may:

Be redundant
Be highly correlated
Contain less useful information

PCA tries to:
✅ Compress data into fewer dimensions
while preserving:
✅ Maximum important information.

WHY IS PCA CALLED “UNSUPERVISED”?

Because:
❌ PCA does NOT use labels/output classes.

It only analyzes:
✅ Input data patterns.

EXAMPLE

Suppose student dataset contains:

Height	Weight	BMI	Age

Here:

Height, Weight, BMI are related.

PCA identifies:
✅ Redundant relationships.

Then compresses them into:

Smaller number of dimensions.
PART 2 — WHY PCA IS NEEDED

Modern datasets may contain:
❌ Hundreds/thousands of features.

Problems:

High computation
Overfitting
Noise
Memory usage

This is called:
✅ Curse of Dimensionality.

PCA helps solve this.

MAIN PURPOSES OF PCA

✅ Dimensionality Reduction
✅ Noise Removal
✅ Faster Training
✅ Better Visualization
✅ Remove Redundant Features

PART 3 — IMPORTANT CONCEPT OF PCA

PCA finds:
✅ New directions/components

called:
✅ Principal Components

that preserve:
✅ Maximum variance/information.

WHAT IS VARIANCE?

Variance means:

How much data spreads.

Higher variance:
✅ More important information.

PCA keeps:
✅ Directions with maximum variance.

PART 4 — PRINCIPAL COMPONENTS

PCA creates:

New artificial axes.

These are:
✅ Principal Components (PCs)

FIRST PRINCIPAL COMPONENT (PC1)

Contains:
✅ Maximum variance.

SECOND PRINCIPAL COMPONENT (PC2)

Contains:
✅ Next highest variance.

Also:

Orthogonal to PC1.
IMPORTANT UNDERSTANDING

PCA rotates the dataset into:
✅ Better informative directions.

### Importing the libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd 
```

### Importing the dataset

```python
dataset=pd.read_csv('Wine.csv')

X = dataset.iloc[:, :-1].values # taken alll the rows

Y=dataset.iloc[:,-1].values
```

### splitting the dataset into thr training set and train set 

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2, random_state=0)
```

### feature scaling 

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()

X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

### applying PCA

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)

X_train = pca.fit_transform(X_train)
X_test = pca.transform(X_test)
```

### training the logistic regression model on the training set 

```python
from sklearn.linear_model import LogisticRegression

classifier = LogisticRegression(random_state=0)

classifier.fit(X_train, Y_train)
```

### Making the confusion matrix 

```python
from sklearn.metrics import confusion_matrix, accuracy_score

Y_pred = classifier.predict(X_test)

cm = confusion_matrix(Y_test, Y_pred)
print(cm)

accuracy = accuracy_score(Y_test, Y_pred)
print(accuracy)
```

### visualizing the training set results 

```python
from matplotlib.colors import ListedColormap
import matplotlib.pyplot as plt
import numpy as np

# Training data
X_set, Y_set = X_train, Y_train

# Create mesh grid
X1, X2 = np.meshgrid(
    np.arange(start=X_set[:, 0].min() - 1,
              stop=X_set[:, 0].max() + 1,
              step=0.01),
    
    np.arange(start=X_set[:, 1].min() - 1,
              stop=X_set[:, 1].max() + 1,
              step=0.01)
)

# Predict for every point in mesh grid
Z = classifier.predict(np.array([X1.ravel(), X2.ravel()]).T)

# Reshape output
Z = Z.reshape(X1.shape)

# Plot decision boundary
plt.contourf(
    X1,
    X2,
    Z,
    alpha=0.75,
    cmap=ListedColormap(('red', 'green'))
)

# Axis limits
plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())

# Plot training points
for i, j in enumerate(np.unique(Y_set)):
    plt.scatter(
        X_set[Y_set == j, 0],
        X_set[Y_set == j, 1],
        c=ListedColormap(('red', 'green'))(i),
        edgecolor='black',
        s=20,
        label=j
    )

# Labels and title
plt.title('Logistic Regression Training Set')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.legend()
plt.show()
```

### visualizing the test set 

```python
from matplotlib.colors import ListedColormap
import matplotlib.pyplot as plt
import numpy as np

# Test data
X_set, Y_set = X_test, Y_test

# Create mesh grid
X1, X2 = np.meshgrid(
    np.arange(start=X_set[:, 0].min() - 1,
              stop=X_set[:, 0].max() + 1,
              step=0.01),

    np.arange(start=X_set[:, 1].min() - 1,
              stop=X_set[:, 1].max() + 1,
              step=0.01)
)

# Predict values for mesh grid
Z = classifier.predict(np.array([X1.ravel(), X2.ravel()]).T)

# Reshape predictions
Z = Z.reshape(X1.shape)

# Plot decision boundary
plt.contourf(
    X1,
    X2,
    Z,
    alpha=0.75,
    cmap=ListedColormap(('red', 'green'))
)

# Set axis limits
plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())

# Plot test points
for i, j in enumerate(np.unique(Y_set)):
    plt.scatter(
        X_set[Y_set == j, 0],
        X_set[Y_set == j, 1],
        c=ListedColormap(('red', 'green'))(i),
        edgecolor='black',
        s=20,
        label=j
    )

# Graph details
plt.title('Logistic Regression Test Set')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.legend()
plt.show()
```
