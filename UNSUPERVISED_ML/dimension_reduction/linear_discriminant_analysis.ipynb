PART 1 — WHAT IS LDA?
SIMPLE DEFINITION

Linear Discriminant Analysis (LDA) is a supervised dimensionality reduction technique that transforms data into a lower-dimensional space while maximizing the separation between different classes.

MAIN IDEA OF LDA

LDA tries to:
✅ Separate classes as much as possible.

while:
✅ Reducing dimensions.

IMPORTANT UNDERSTANDING

PCA focuses on:
✅ Maximum variance

LDA focuses on:
✅ Maximum class separability

This is the biggest concept in LDA.

WHY IS LDA CALLED “SUPERVISED”?

Because:
✅ LDA uses class labels/output labels.

Example:

Student Marks	Class
90	Pass
30	Fail

LDA knows:

Which points belong to which class.
PART 2 — WHY LDA IS NEEDED

Suppose dataset contains:

Many features
Multiple classes

Problems:

❌ High computation

LDA solves this by:
✅ Finding best feature directions that separate classes.

MAIN GOAL OF LDA

LDA tries to:

while
✅ Minimizing spread within same class.

VERY IMPORTANT CONCEPT

LDA wants:

Class A   ← far apart →   Class B

and also:

Same-class points close together

Avoid LDA when:

PART 12 — IMPORTANT ASSUMPTIONS OF LDA

LDA assumes:

1. DATA IS LINEARLY SEPARABLE

LDA mainly works for:
✅ Linear class separation.

2. FEATURES FOLLOW NORMAL DISTRIBUTION

LDA assumes:

Gaussian distribution.
3. SAME COVARIANCE ACROSS CLASSES

All classes assumed to have:

Similar covariance.

### Importing the libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd 
```

### Importing the dataset

```python
dataset=pd.read_csv('Wine.csv')

X = dataset.iloc[:, :-1].values # taken alll the rows

Y=dataset.iloc[:,-1].values
```

### splitting the dataset into thr training set and train set 

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2, random_state=0)
```

### feature scaling 

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()

X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

### applying PCA

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis as LDA
lda = LDA(n_components=2)# numbers of features to be extracted
X_train= lda.fit_transform(X_train,Y_train) # lda requires features as well as dependent varaible 
X_test = lda.transform(X_test)
```

### training the logistic regression model on the training set 

```python
from sklearn.linear_model import LogisticRegression

classifier = LogisticRegression(random_state=0)

classifier.fit(X_train, Y_train)
```

### Making the confusion matrix 

```python
from sklearn.metrics import confusion_matrix, accuracy_score

Y_pred = classifier.predict(X_test)

cm = confusion_matrix(Y_test, Y_pred)
print(cm)

accuracy = accuracy_score(Y_test, Y_pred)
print(accuracy)
```

### visualizing the training set results 

```python
from matplotlib.colors import ListedColormap
import matplotlib.pyplot as plt
import numpy as np

# Training data
X_set, Y_set = X_train, Y_train

# Create mesh grid
X1, X2 = np.meshgrid(
    np.arange(start=X_set[:, 0].min() - 1,
              stop=X_set[:, 0].max() + 1,
              step=0.01),
    
    np.arange(start=X_set[:, 1].min() - 1,
              stop=X_set[:, 1].max() + 1,
              step=0.01)
)

# Predict for every point in mesh grid
Z = classifier.predict(np.array([X1.ravel(), X2.ravel()]).T)

# Reshape output
Z = Z.reshape(X1.shape)

# Plot decision boundary
plt.contourf(
    X1,
    X2,
    Z,
    alpha=0.75,
    cmap=ListedColormap(('red', 'green'))
)

# Axis limits
plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())

# Plot training points
for i, j in enumerate(np.unique(Y_set)):
    plt.scatter(
        X_set[Y_set == j, 0],
        X_set[Y_set == j, 1],
        c=ListedColormap(('red', 'green'))(i),
        edgecolor='black',
        s=20,
        label=j
    )

# Labels and title
plt.title('Logistic Regression Training Set')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.legend()
plt.show()
```

### visualizing the test set 

```python
from matplotlib.colors import ListedColormap
import matplotlib.pyplot as plt
import numpy as np

# Test data
X_set, Y_set = X_test, Y_test

# Create mesh grid
X1, X2 = np.meshgrid(
    np.arange(start=X_set[:, 0].min() - 1,
              stop=X_set[:, 0].max() + 1,
              step=0.01),

    np.arange(start=X_set[:, 1].min() - 1,
              stop=X_set[:, 1].max() + 1,
              step=0.01)
)

# Predict values for mesh grid
Z = classifier.predict(np.array([X1.ravel(), X2.ravel()]).T)

# Reshape predictions
Z = Z.reshape(X1.shape)

# Plot decision boundary
plt.contourf(
    X1,
    X2,
    Z,
    alpha=0.75,
    cmap=ListedColormap(('red', 'green'))
)

# Set axis limits
plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())

# Plot test points
for i, j in enumerate(np.unique(Y_set)):
    plt.scatter(
        X_set[Y_set == j, 0],
        X_set[Y_set == j, 1],
        c=ListedColormap(('red', 'green'))(i),
        edgecolor='black',
        s=20,
        label=j
    )

# Graph details
plt.title('Logistic Regression Test Set')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.legend()
plt.show()
```
