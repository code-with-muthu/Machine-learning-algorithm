# k- MEANS CLUSTERING

### DEF:
       K-Means Clustering is an unsupervised machine learning algorithm used to divide data into K number of clusters based on similarity.

       The algorithm groups similar data points together and separates dissimilar points into different clusters.

### WORKING
  Step 1: Choose K clusters
  Suppose:
                    K=3
  This means the algorithm will create 3 clusters.

   Step 2: Initialize Centroids
   Randomly choose K points as initial centroids.

   Step 3: Calculate Distance

       For every data point, calculate the distance from each centroid.
       Usually Euclidean distance is used:

              d = squrt (x2 - x1)^2 + (y2 - y1)^2
	​The point is assigned to the nearest centroid.

    Step 4: Form Clusters
   All nearby points are grouped into the same cluster.

Step 5: Update Centroids
 Compute the new centroid by taking the mean of all points in the cluster.
                             μ=∑xi​ / n​​


where:

μ = centroid
xi = data points
n = number of points
   
Step 6: Repeat
         Repeat Steps 3–5 until:
         centroids stop changing, 
         clusters become stable

This is called convergence.

Example

Suppose a shopping company has customer data:

Age
Salary
Spending score

K-Means can group customers into:

1. High spenders
2. Medium spenders
3. Low spenders

without predefined labels.

### Advantages
1. Simple and fast
2. Easy to implement
3. Works well for large datasets
4. Efficient for customer segmentation
### Disadvantages
1. Need to choose K value manually
2. Sensitive to outliers
3. Different initial centroids may give different results
4. Works best for spherical clusters

##  ELBOW METHOD:

### DEF :
The Elbow Method is a technique used in K-Means Clustering to find the optimal value of K (number of clusters).

It helps determine how many clusters should be created for the dataset.

### Why is Elbow Method Needed?

In K-Means, we must choose the value of:

             K=Number of clusters
But choosing K randomly may produce poor clustering.
The Elbow Method helps select the best K value.

Main Concept

K-Means tries to minimize the Within-Cluster Sum of Squares (WCSS).

WCSS means:

the sum of squared distances between data points and their nearest centroid.

### Formula:

WCSS=∑i=1K​∑x∈Ci​​(x−μi​)^2
	​
Ci= cluster
μi = centroid of cluster
x = data point

Lower WCSS means better clustering.

### How Elbow Method Works

Step 1: Choose Different K Values

Run K-Means for different values of K.

Example:

K=1,2,3,4,5,6,7...
Step 2: Compute WCSS

For each K value:

perform K-Means clustering
calculate WCSS
Step 3: Plot Graph

Draw a graph:

X-axis → Number of clusters (K)
Y-axis → WCSS
Step 4: Find the “Elbow Point”

As K increases:

WCSS decreases

Initially:

WCSS drops sharply

After some point:

reduction becomes slow

The point where the curve bends like an elbow is called the Elbow Point.

That K value is considered optimal.

Why is it Called “Elbow”?

Because the graph looks like a human arm bend.

Example trend:

K	WCSS
1	1000
2	600
3	300
4	200
5	180
6	170

Here, the major reduction stops after K = 3.

So:

K=3

is chosen.

Simple Intuition
Small K → clusters too broad
Very large K → too many unnecessary clusters

Elbow Method finds the balance point.

This helps:

improve clustering accuracy
reduce computation time
